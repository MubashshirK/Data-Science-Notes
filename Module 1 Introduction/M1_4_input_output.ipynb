## **Python Output**

We use the print( ) function to output data to the standard output device.

```python
print("Hello, World!")
```

```python
a = 10
print("The value of a is", a)        # Python 3
print("The value of a is " + str(a))
```

## **Output Formatting**

```python
a = 10; b = 20    # multiple statements in single line.

print("The balue of a is {} and b is {}".format(a, b))    #default
```

```python
a = 10; b = 20    # multiple statements in single line.

print("The value of b is {1} and a is {0}".format(a, b))    # specify position
```

```python
# we can use keyword arguments to format the string
print("Hello {name}, {greeting}".format(name="Kyle", greeting="Good Morning!"))
```

```python
# we can combine positional arguments with keyword arguments
print('The story of {0}, {1}, and {other}'.format('Bill', 'Manfred', other='Georg'))
```

## **Python Input**

Want to take the input from the user?<br> In Python, we have the input( ) function to allow this.
