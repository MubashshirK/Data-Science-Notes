# **Variables**

A variable is a location in memory used to store some data (value).

They are given unique names ot differentiate between different memory locations. The rules for writing a variable name is same as the rule for writing identifiers in Python.

We don't need to declare a variable before using it. In Python, we simply assign a value to a variable and it will exist. We don't even have to declare the type of the variable. This is handled internally according to the type of the value we assign to the vairable.

# **Variable Assignments**

```python
#We use the assignment operator (=) to assign values to a variable

a = 10    # int
b = 5.5    # float
c = "ML"    # string
```

# **Multiple Assignments**

```python
a, b, c = 10, 5.5, "ML"
```

```python
a = b = c = "AI"    # Assign the same value to multiple variable at once
```

# **Data Types**

Every value in Python has a datatype. Since everything is an object in Python programming, data types are actually classes and variables are instance (object) of these classes.

# **Numbers**

Integers, floating point numbers and complex numbers falls under Python numbers category. They are defined as int, float and complex class in Python.

We can use the type() function to know which class a variable or a value belongs to and the isinstance() function to check if an object belongs to a particular class.

```python
a = 5                              # Data type is implicitly set to integer
print(a, "is of type", type(a))
```

```python
a = 5.5                            # Data type is changed to float
print(a, "is of type", type(a))
```

```python
a = 1 + 2j                         # Data type is changed to complex number
print(a, "is complex number?")
print(isinstance(1+2j, complex))
print(type(a))
```

# **Boolean**

Boolean represents the truth values False and True.

```python
a = True
print(type(a))
```

# **Python Strings**

String is sequence of Unicode characters.

We can use single quotes or double quotes to represent strings.

Multi-line strings can be denoted using triple quotes, " or """.

A string in Python consists of a series or sequence of characters - letters, numbers, and special
3characters.

Strings can be indexed - often synonymously called subscripted as well.

Similar to C, the first character of a string has the index 0.

```python
s = "This is Online AI course"
print(s)
print(type(s))
```

```python
print(s[0])    # Last char s[len(s)-1] or s[-1]
```

```python
# Slicing
s[5:10]        # Prints from index 5 to 9
```

# **Python List**

List is an ordered sequence of items. It is one of the most used datatype in Python and is very flexiable. All the items in a list do not need to be of the same type.

Declaring a list is, Items separated by commas are enclosed within brackets [ ].

```python
a = [10, 20.5, 'Hello']
print(a[1])                # print 1st index element
```

Lists are mutable, meaning, value of elements of a list can be altered.

```python
a[1] = 30.7
print(a)
```

# **Python Tuple**

Tuple is an ordered sequence of items same as list. The only difference is the tuples are immutable.
Tuples once created cannot be modified.

```python
t = (1, 1.5, "ML")
```

```python
print(t[1])    # Extract particular element
```

```python
t[1] = 1.25
```

# **Python Set**

Set is an unordered collection of unique items. Set is defined by values separated by comma inside braces { }. Items in a set are not ordered.

```python
a = {10, 30, 20, 40, 5}
print(a)
```

```python
print(type(a))
```

>We can perform set operations like union, intersection on two sets. Set have unique values.

```python
s = {10, 20, 20, 30, 30, 30}
print(s)                      # Automatically set won't consider duplicate elements
```

```python
print(s[1])    # we can't print particular element in set because
               # it's unordered collections of items
```

# **Python Dictionary**

Dictionary is an unordered collectino of key-value pairs.

In Python, dictionaries are defined within braces { } with each item being a pair in the form key : value.
<br>Key and value can be of any type.

```python
d = {'a': 'apple', 'b': 'bat'}
print(d['a'])
```
