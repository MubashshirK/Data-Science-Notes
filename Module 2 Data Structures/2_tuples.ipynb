# **Tuples**

- A tuple is similar to list.

- The difference between the two is that we can't change the elements of tuple once it is assigned whereas in the list, elements can be changed.

## **Tuple creation**

```python
#empty tuple
t = ()

#tuple having integers
t = (1, 2, 3)
print(t)

#tuple with mixed datatypes
t = (1, 'kyle', 28, 'abc')
print(t)

#nested tuple
t = (1, (2, 3, 4), [1, 'kyle', 28, 'abc'])
print(t)
```

```python
#only parenthesis is not enough
t = ('kyle')
type(t)
```

```python
#need a comma at the end
t = ('kyle',)
type(t)
```

```python
#parenthesis is optional
t = 'kyle',
print(type(t))

print(t)
```

## **Accessing Elements in Tupe**

```python
t = ('kyle', 'john', 'joel', 'ellie', 'kratos', 'caraxes')

print(t[1])
```

```python
#negative index
print(t[-1])
```

```python
#nested tuple
t = ('ABC', ('kyle', 'joel', 'ellie'))

print(t[1])
```

```python
#Slicing
t = (1, 2, 3, 4, 5, 6)

print(t[1:4])

#print elements from starting to 2nd last elements
print(t[:-2])

#print elements from starting to end
print(t[:])
```

## **Changing a Tuple**

- Unlike lists, tuple are immutable.

> This means that elements of a tuple cannot be changed once it has been assigned. But, if the element is itself a mutable datatype like list, its nested items can be changed.

```python
#creating tuple
t = (1, 2, 3, 4, [5, 6, 7])
t[2] = 'x'    #will get TypeError
```

```python
t[4][1] = 'kyle'
print(t)
```

```python
#concatinating tuples
t = (1, 2, 3) + (4, 5, 6)
print(t)
```

```python
#repeat the elements in a tuple for a given number of times using the * operator
t = (('kyle',) * 4)
print(t)
```

## **Tuple Deletion**

```python
#We cannot change the elements of a tuple.
#That also means we cannot delete or remove items from a tuple.

#delete entire tuple using del keyword
t = (1, 2, 3, 4, 5, 6)

#delete entire tuple
del t
```

## **Tuple Count**

```python
t = (1, 2, 3, 1, 3, 3, 4, 1)

#get the frequency of particular element appears in a tuple
t.count(1)
```

## **Tuple Index**

```python
t = (1, 2, 3, 1, 3, 3, 4, 1)

print(t.index(3))    #return index of the first element is equal to 3

#print index of the 1
```

## **Tuple Membership**

```python
#test if an item exists in a tuple or not, using the keyword in.
t = (1, 2, 3, 4, 5, 6)

print(1 in t)
```

```python
print(7 in t)
```

## **Built in Functions**

## **Tuple Length**

```python
t = (1, 2, 3, 4, 5, 6)
print(len(t))
```

## **Tuple Sort**

```python
t = (4, 5, 1, 3, 2)

new_t = sorted(t)
print(new_t)    #Take elements in the tuple and return a new sorted list
                # (does not sort the tuple iteslf).
```

```python
# get the largest element in a tuple
t = (2, 5, 1, 6, 9)

print(max(t))
```

```python
#get the smallest element in a tuple
print(min(t))
```

```python
#get sum of elements in the tuple
print(sum(t))
```
